```python
import pandas as pd
import zipfile
import urllib.request
from io import BytesIO

url = "https://archive.ics.uci.edu/static/public/228/sms+spam+collection.zip"

response = urllib.request.urlopen(url)

with zipfile.ZipFile(BytesIO(response.read())) as z:
    filename = [f for f in z.namelist()
                if f.endswith("SMSSpamCollection")][0]

    df = pd.read_csv(
        z.open(filename),
        sep="\t",
        header=None,
        names=["labels", "emails"]
    )

emails = df["emails"]
labels = df["labels"]

print(df.shape)
df.head()
```
